# Qwen (Ollama)

`1_3_ollama.ipynb`와 동일한 구조이며, 모델만 Qwen으로 바꾼 예제입니다.

실행 전 터미널에서 모델을 먼저 받아야 합니다:

```bash
ollama pull qwen2.5:3b
```

OpenAI 호환 엔드포인트를 쓰기 때문에 코드는 OpenAI / Gemini 예제와 똑같습니다.
바뀌는 것은 `base_url`과 `model` 두 가지뿐입니다.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
````

````python
load_dotenv(override=True)
````

````python
ollama_url = os.getenv('OLLAMA_URL')
print(ollama_url)
````

````python
qwen = OpenAI(api_key="ollama", base_url=ollama_url)
````

````python
system_prompt = """
당신은 IT 보안 전문가입니다. 
시스템 보안에 대한 전문 지식을 제공하며 새롭게 발생하는 보안 문제에 대해서도 대처 방법을 제공합니다
"""
````

````python
user_prompt = """
나는 IT 회사의 보안 책임자입니다.
현재 가장 대두되고 있는 보안 이슈는 무엇이 있나요?

"""
````

````python
message = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
````

````python
response = qwen.chat.completions.create(model="qwen2.5:3b", messages=message)
````

````python
import json

print(json.dumps(response.model_dump(), indent=2, ensure_ascii=False))
````

````python
print(response.choices[0].message.content)
````
